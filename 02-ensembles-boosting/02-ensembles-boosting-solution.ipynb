# Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier
from sklearn.metrics import classification_report, accuracy_score
```

### Today's Data
We're going to be using the famous spam detection dataset. It's documented in UCI's ML database [here](https://archive.ics.uci.edu/ml/datasets/spambase) and is used frequently throughout the ESL book for tree-based models.

![](../assets//spam.png)

```python
# Load the data
names_raw = open('../data/spam-colnames.txt', 'r').readlines()
names = [line.split(':')[0] for line in names_raw[33:]] + ['is_spam']
spam = pd.read_csv('../data/spam.csv', sep=r'\s+', names=names, engine='python')
spam.head()
```

```python
# Assumption: Data are very clean! No missings, categorical features, etc.
# Let's move along here for the sake of time.
```

```python
X = spam.drop(columns=['is_spam'])
y = spam['is_spam']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=1234)
```

### Let's see how a Decision Tree Classifier performs without any boosting

```python
tree = DecisionTreeClassifier(max_depth=1, random_state=111)
tree.fit(X_train, y_train)
```

```python
print(tree.score(X_train, y_train))
print(tree.score(X_test, y_test))
```

# The AdaBoost model
---
`scikit-learn` implements the `SAMME` algorithm, not `AdaBoost.M1` like we learned in class, but it's very similar.

[Docs here!](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.AdaBoostClassifier.html)

```python
# Let's take a peek at the docs...
AdaBoostClassifier()
```

```python
# Try max_depth=None vs max_depth=1
tree = DecisionTreeClassifier(max_depth=1, random_state=111)
ada = AdaBoostClassifier(tree, n_estimators=100, learning_rate=1)
```

```python
ada.fit(X_train, y_train)
```

```python
print(ada.score(X_train, y_train))
print(ada.score(X_test, y_test))
```

# Now You: Tune the model

Try the following:

1. Increase `n_estimators` from 50 to 200 — does accuracy improve?
2. Decrease the `learning_rate` — does it overfit less?
3. Try deeper trees (e.g., `max_depth=2`) — what happens to precision and recall?
4. Compare with a simple DecisionTreeClassifier on its own — is AdaBoost better?

Use the cell below to experiment!

```python
# Your turn: tune these parameters
model = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth=2),
    n_estimators=100,
    learning_rate=0.5,
    random_state=42
)
model.fit(X_train, y_train)

# Evaluate
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

# The GradientBoost model
---
Remember the strategy and impetus behind gradient boosting is similar to AdaBoosting with a few differences. The main one is that you can choose your loss function! The `scikit-learn` implementation already pre-incorporates the tree model, so we get to skip a step there.

[Docs here!](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.GradientBoostingClassifier.html#sklearn.ensemble.GradientBoostingClassifier)

```python
# Instantiate and fit
gb = GradientBoostingClassifier()
gb.fit(X_train, y_train)
```

```python
# Easy peasy - didn't even mess with tuning params!
print(gb.score(X_train, y_train))
print(gb.score(X_test, y_test))
```

### Let's GridSearch!

```python
pgrid = {
    'learning_rate': [0.1, 1, 10],
    'n_estimators': [10, 100],
    'max_depth': [None, 1, 2, 3]
}

gb = GradientBoostingClassifier(random_state=222)
gs = GridSearchCV(gb, pgrid, cv=5, n_jobs=6)
```

```python
%%time
gs.fit(X_train, y_train)
```

```python
print(gs.score(X_train, y_train))
print(gs.score(X_test, y_test))
```

```python
gs.best_params_
```

# Conclusions and Takeaways
---

* Boosting is a very powerful technique that is **best used to improve the bias of weak learners.** 
* Boosting reduces bias by improving weak learners, contrast this with bagging with reduces variance by average too-strong (overfit) learners.
* You can mess with the tuning parameters if you'd like, but generally if you train for long enough, the learning rate won't matter.

### AdaBoost vs GradientBoost vs XGBoost
- AdaBoost is about reweighting the preceding model's errors in subsequent iterations.
- Gradient boosting is about fitting subsequent models to the residuals of the last model.
- XGBoost is based on gradient boosting but is optimized to work on large data with missing values.

### Model Boosting Pros
* Top-tier performance
* You get all the same benefits associated with tree-based models

### Model Boosting Cons
* All hope of interpretation, gone!
* Computational complexity is the highest we've ever seen - can cause models to fit slowly!
* Because of serial nature of algorithm, parallelization is impossible.
